```python
import numpy as np

# 1. Save grid and projection info to a NetCDF
def save_grid_netcdf(ds, out_path="grid_info.nc"):
    # Extract only lat, lon, and projection attributes if present
    grid_ds = xr.Dataset()
    for coord in ['lat', 'lon']:
        if coord in ds.coords:
            grid_ds[coord] = ds[coord]
    # Copy projection info if available
    for k, v in ds.attrs.items():
        if 'proj' in k.lower() or 'crs' in k.lower():
            grid_ds.attrs[k] = v
    grid_ds.to_netcdf(out_path)

# Example usage:
save_grid_netcdf(ds)
```
