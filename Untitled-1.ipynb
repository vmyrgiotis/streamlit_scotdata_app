```python
# Align t2m/tp valid_time to existing _D.time (LC_Type5 timeline: 2001..2025)
# _D.time may be CFTime (e.g., DatetimeJulian), so avoid direct pd.to_datetime on raw values
time_index = _D.indexes.get("time", None)
if hasattr(time_index, "to_datetimeindex"):
    target_time = time_index.to_datetimeindex()
else:
    target_time = pd.DatetimeIndex([
        datetime(t.year, t.month, t.day, t.hour, t.minute, t.second)
        for t in _D["time"].values
    ])
valid_time = pd.to_datetime(_D["valid_time"].values)  # length 26

for v in ["t2m", "tp"]:
    if v in _D and "valid_time" in _D[v].dims:
        da = _D[v].rename({"valid_time": "time"})

        rename_map = {
            k: v2 for k, v2 in {
                "latitude": "lat",
                "longitude": "lon",
                "y": "lat",
                "x": "lon",
            }.items() if k in da.dims
        }
        if rename_map:
            da = da.rename(rename_map)

        # attach full valid_time first, then align to _D.time (drops year 2000)
        da = da.assign_coords(time=("time", valid_time))
        da = da.sel(time=target_time).transpose("time", "lat", "lon")

        _D[v] = da

# keep _D.time as-is (25), just remove helper coord
# _D = _D.drop_vars("valid_time", errors="ignore")
```

```python
vegetation_codes = [1, 2, 3, 4, 5, 7, 8]

# Keep Npp_500m only where LC_Type5 is a vegetation class; else set to NaN
_D["Npp_500m"] = _D["Npp_500m"].where(_D["LC_Type5"].isin(vegetation_codes))
_D["ocs_0_100cm"] = _D["ocs_0_100cm"].where(_D["ocs_0_100cm"]>0)
```
